# 2D line plot

2d line plot is used in bivariate analysis(2 columns)

on which type of data we should use it
-categorical-> numerical
and numerical->numerical

use-case: Time series data(1 quantity ko time access per measure kerte ho)

mostly very good in using on data when one column is numerical and other is categorical.

```python
# plotting a simple function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

# plt.plt(x_axis=categorical-column, y_axis=numerical-column)
plt.plot(year,price)
```

```python
# from a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# plotting multiple plots together
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# label title
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit sharma vs V kohli')
plt.xlabel('seasons')
plt.ylabel('runs_scored')
```

```python
# color(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['V Kohli'], color='green')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black')
plt.title("Rohit sharma vs virat kholi season and runs")
plt.xlabel('season')
plt.ylabel('runs')
```

```python
# we have a hidden parameter here called the linestyle
plt.plot(batsman['index'], batsman['V Kohli'], color='yellow', linestyle="dotted")
plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle="dashed")
plt.title("rohit-sharma vs virat kholi season and runs")
plt.xlabel('season')
plt.ylabel('runs')
```

```python
# we can also change the linwidth with the same hidden parameter
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='dashdot', linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'], color='yellow', linestyle='dashed', linewidth=1)
```

```python
plt.plot(batsman['index'], batsman['RG Sharma'], color='green', linestyle='dotted', linewidth=2, marker='D')
plt.plot(batsman['index'], batsman['V Kohli'], color='blue', linestyle='dashed', linewidth=3, marker='o')
```

```python
# legend and locations
plt.plot(batsman['index'], batsman['V Kohli'], color='blue', linestyle='dashdot', linewidth=3, marker='D', markersize=9, label='virat-kholi')
plt.plot(batsman['index'], batsman['RG Sharma'], color='orange', linestyle='solid', linewidth=2, marker='o', markersize=5, label='rohit-sharma')

# we cant see label cuz we also need to call legend
# plt.legend()
# by default it actomatically finds the right position in the graph
# but you can also change it according to your need
plt.legend(loc='center')
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000, 4500000]
year = [2015,2016,2017,2018,2019,2020, 2021]

# plt.plt(x_axis=categorical-column, y_axis=numerical-column)
plt.plot(year,price)
# now  due to 2021 value which is a outlier your graph variation would not be proper
```

```python
# now in these type of cases we can use limiting axes
# limiting axes
price = [48000,54000,57000,49000,47000,45000, 4500000]
year = [2015,2016,2017,2018,2019,2020, 2021]

# plt.plt(x_axis=categorical-column, y_axis=numerical-column)
plt.plot(year,price)
# now  due to 2021 value which is a outlier your graph variation would not be proper

# for these kind of senarios we use limiting feature
plt.xlim(2017, 2021)
plt.ylim(40000, 75000)
```

```python
# so lets put the grid in our previous data
plt.plot(batsman['index'], batsman['V Kohli'], color='green', linestyle='dashdot', linewidth=5, marker='D', markersize=9, label='virat-kohili')
plt.plot(batsman['index'], batsman['RG Sharma'], color='yellow', linestyle='dashed', linewidth=3, marker='>', markersize=5, label='rohit-sharma')

plt.title("rohit-sharma vs virat kholini seasonal runs over the year")

plt.xlabel('season')
plt.ylabel('run')

plt.legend()

plt.grid()

# there also show() function which are applied to different text editors to
# explicitely show
plt.show()
```
