```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#Colored Scatterplots

```python
iris = pd.read_csv('iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa': 1, 'Iris-virginica':2, 'Iris-versicolor':3})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='jet', alpha=0.7)
# alpha = opaque-city
plt.xlabel("sepal_length")
plt.ylabel("petal_length")
plt.colorbar()
plt.grid()
plt.show()
```

# Increasing the Size of the graph

```python
plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'], iris['PetalLengthCm'], c=iris['Species'], cmap='summer')
plt.xlabel("sepal_length")
plt.ylabel("petal_length")
plt.colorbar()
plt.show()
```

# Annotation

```python
# we will use the batters dataset
batters = pd.read_csv('batter.csv')
```

```python
batters.shape
```

```python
batters.head(5)
```

```python
batters.shape
```

```python
sample_df = batters.head(100).sample(25,random_state=5)
```

```python
sample_df
```

```python
# for these kinds of graph only we will amke the fig size bigger
plt.figure(figsize=(15,7))

plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])


plt.grid()
plt.show()
```

```python
# you can name these points
x = [1,2,3,4,5]
y = [5,2,4,2,1]

plt.scatter(x,y)
plt.text(1,5,'first point')
plt.text(2,2,'second point')
plt.text(3,4,'third point')
plt.text(4,2,'fourth point')
plt.text(5,1,'fifth point', fontdict=({'size':12,'color':'yellow'}))
plt.grid()
plt.show()
```

# Vertical /Horizontal line to show for the bats man with more and less than 130

```python
plt.figure(figsize=(15,7))

plt.scatter(sample_df['avg'], sample_df['strike_rate'], s=sample_df['runs'])

# for horizontal line we will write
plt.axhline(135,color='green')

# for vertical line we will write
plt.axvline(25, color='yellow')

# annotate = naming the points
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i], sample_df['strike_rate'].values[i], sample_df['batter'].values[i])

# plt.grid()
plt.xlabel('average')
plt.ylabel('strike_rate')
plt.show()
```

# Subplots

```python
#  a diff way to plot a graph
```

```python
batters.head()
```

```python
# for increasing the size
plt.figure(figsize=(15,7))

plt.scatter(batters['avg'], batters['strike_rate'], marker='+', color='orange')
plt.title("scatter_plot_of_batters")
plt.xlabel('average')
plt.ylabel('strike_rate')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15,7))  #here we called the subplot using plt and got ax and fig as objects

# using object ax we called scatter
ax.scatter(batters['avg'], batters['strike_rate'], marker='+', color='red')
# using the ax object we call set_title()
ax.set_title("scatter_plot_of_batters")
ax.set_xlabel('average')
ax.set_ylabel('strike rate')
fig.show()
```

```python
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10,6))
# in this we have single figure and 2 axies
# first axis = first graph
ax[0].scatter(batters['avg'], batters['strike_rate'], color='green', marker="+")

# seconf axis = second graph
ax[1].scatter(batters['avg'], batters['runs'], color='yellow', marker="*")


# we can set the first graphs labels
ax[0].set_title("scatter plot avg and strike_rate")
ax[0].set_ylabel("strike_rate")

ax[1].set_title("scatter plot avg and runs")
ax[1].set_ylabel("runs")
```

```python
# we will define a figure
fig = plt.figure(figsize=(15,7))

# now we will cretae the projetion
ax = plt.subplot(projection='3d')

ax.scatter( batters['runs'],batters['avg'], batters['strike_rate'], marker="+", color='Red')
ax.set_title("IPL  batsman analysis")
ax.set_xlabel("runs")
ax.set_ylabel("avg")
ax.set_zlabel("strike_rate")
```
